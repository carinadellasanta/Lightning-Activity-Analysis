### 2.2 Dataset Separation and File Conversion

I build a function that loads .mat files, converts them into pandas DataFrames and saves them as .csv files for easier data analysis.

```python
def process_mat_file(file, input_path, output_path):
    
    full_path = os.path.join(input_path, file)
    
    data = loadmat(full_path)
    
    key = [k for k in data.keys() if not k.startswith("__")][0]
    array = data[key]
    
    df = pd.DataFrame(array)
    
    name = os.path.splitext(file)[0]
    
    df.to_csv(os.path.join(output_path, f"{name}.csv"), index=False)
```

I separate the dataset into two categories: 

- Events: individual lightning strike records
- Daily: total number of lightning strikes per day

```python
files = [f for f in all_files if "suma" not in f]
Sfiles = [f for f in all_files if "suma" in f]

print('files: ', files)
print('Sfiles: ', Sfiles)

output_path = os.path.join(base_path,'Data','raw',"Events_csv")
os.makedirs(output_path, exist_ok=True)

Soutput_path = os.path.join(base_path,'Data','raw', "Daily_csv")
os.makedirs(Soutput_path, exist_ok=True)
```

```python
for file in files:
    process_mat_file(file, input_path, output_path)

for file in Sfiles:
    process_mat_file(file, input_path, Soutput_path)
```
